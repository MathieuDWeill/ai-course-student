# Bloc 3.7 — Final data product recommendation

**Decision problem:** what should the organization do next based on the full evidence chain?

Output: final recommendation note.

```python
from pathlib import Path
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd()
SNAPSHOT = "iot_FR_today_5-y_chatgpt_iphone_meteo.csv"
for candidate in [ROOT, *ROOT.parents]:
    if (candidate / "data" / "snapshots" / SNAPSHOT).exists():
        ROOT = candidate
        break
OUT = ROOT / "outputs"
for d in [OUT / "bloc1", OUT / "bloc2", OUT / "bloc3", OUT / "final_product"]: d.mkdir(parents=True, exist_ok=True)
DATA = ROOT / "data" / "snapshots"
def load_clean_long():
    p=OUT/"bloc1"/"clean_trends_long.csv"
    if p.exists(): return pd.read_csv(p, parse_dates=["date"])
    raw=pd.read_csv(DATA/"iot_FR_today_5-y_chatgpt_iphone_meteo.csv", parse_dates=["date"])
    return raw.melt(id_vars="date", var_name="signal", value_name="interest")
def load_clean_wide():
    p=OUT/"bloc1"/"clean_trends_wide.csv"
    if p.exists(): return pd.read_csv(p, parse_dates=["date"])
    raw=pd.read_csv(DATA/"iot_FR_today_5-y_chatgpt_iphone_meteo.csv", parse_dates=["date"])
    return raw.sort_values("date")
```

```python
kpi=pd.read_csv(OUT/"bloc3"/"kpi_snapshot.csv") if (OUT/"bloc3"/"kpi_snapshot.csv").exists() else pd.DataFrame()
roi=pd.read_csv(OUT/"bloc3"/"prioritization_roi.csv") if (OUT/"bloc3"/"prioritization_roi.csv").exists() else pd.DataFrame()
recommendation="""# Final Recommendation

Adopt a monthly weak-signal review as the first data product.

Why: it is low-effort, uses reproducible public data, and creates a repeatable product decision ritual.

Do next: publish the dashboard dataset, review ChatGPT shifts monthly, and run one small content or discovery experiment when the signal changes materially.
"""
(OUT/"bloc3"/"final_recommendation.md").write_text(recommendation)
print(recommendation)
```
