```python
from pathlib import Path
import os,zipfile,subprocess,sys
candidates=list(Path('/kaggle/input').rglob('experiment.zip'))
assert len(candidates)==1, 'Joindre exactement un experiment.zip comme donnée Kaggle.'
work=Path('/kaggle/working/mllab');work.mkdir(exist_ok=False)
with zipfile.ZipFile(candidates[0]) as z:
    for name in z.namelist():
        p=Path(name)
        assert not p.is_absolute() and '..' not in p.parts, 'Chemin ZIP refusé'
    z.extractall(work)
# Contrat de dépendances : utiliser l’image préinstallée ou préparer les packages autorisés avant la séance.
# Ne pas activer Internet si le concours l’interdit.
subprocess.run([sys.executable,'run.py'],cwd=work,check=True)
print('Résultat à récupérer :',work/'result.zip')
```
